Gerar um Transformador de Tempo, Por exemplo receber 'X' segundos e retornar quantos min são 

```python
#Definição de variáveis
segundos = int(input('Insira quantos segundos deseja converter: '))

#Conversão de segundos para horas, minutos e segundos
minutos = segundos //60
segundos_restantes = segundos % 60
horas = minutos //60
minutos_restantes = minutos % 60
segundos_restantes_final = segundos_restantes % 60  

print(f"{horas} horas, {minutos_restantes} minutos e {segundos_restantes_final} segundos")
```
