Calculadora de IMC com DEF

```python
nome = input("Insira seu nome: ")

# Convertemos para float para permitir cálculos (e tratamos a altura de CM para Metros)
peso = float(input("Informe seu peso em KG: "))
altura_cm = float(input("Informe sua altura em CM: "))
altura = altura_cm / 100  # Convertendo centímetros para metros


def calcular_imc(peso, altura):
    """Calcula o valor numérico do IMC."""
    return peso / (altura ** 2)

def classificar_imc(imc):
    """Retorna a string com a classificação baseada no IMC."""
    if imc < 18.5:
        return "Abaixo do peso"
    elif 18.5 <= imc < 25.0:
        return "Peso normal (Eutrofia)"
    elif 25.0 <= imc < 30.0:
        return "Sobrepeso"
    elif 30.0 <= imc < 34.9:
        return "Obesidade Grau I"
    elif 35.0 <= imc < 39.9:
        return "Obesidade Grau II"
    else:
        return "Obesidade Grau III (Mórbida)"

# --- O que faltava: Chamar as funções e exibir o resultado ---
resultado_imc = calcular_imc(peso, altura)
classificacao = classificar_imc(resultado_imc)

print(f"\n--- Resultado ---")
print(f"Nome: {nome}")
print(f"Seu IMC é: {resultado_imc:.2f}")
print(f"Classificação: {classificacao}")
```
