Dicionários (Dict), são estruturas chave : valor. Mutáveis como a Lista

```python
#Gerando info sobre um clube de futebol

equipe = {
"nome":"Cruzeiro",
"pais":"Brasil",
"value":"130M",
"ano":1921
}
```

```python
#Printando esse dicionário
print(type(equipe))
print(equipe)
```

```python
#Acessando valor específico da chave

print(f"Nome da equipe: {equipe['nome']}")
print(f"Valor de mercado: {equipe.get("value")}")
```

```python
#Adionando nova Chave:Valor
print(equipe)
equipe["alcunha"] = "cabuloso"
print(f"Dict atualizado: {equipe}")
```

```python
#Removendo um valor existente
equipe["alcunha"] = "Cabuloso"
print(equipe)
```
