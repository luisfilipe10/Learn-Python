As tuplas são imutáveis.

mas é possivel converter em lista, realizar a alteração, e após isso, retornar para Tupla


```python
coord = (10.4,11.3)

print(f"coordenada x: {coord[0]}\ncoordenada y: {coord[1]}")
```

```python
coord = (10.4,11.3)
```

```python
type(coord)
coord = list(coord)
print(coord)
type(coord)
```

```python
coord.append(11.55)
print(coord)
print(type(coord))
```

```python
coord = tuple(coord)
print(coord)
print(type(coord))
```

```python
dias = ("segunda","terça","quarta","quinta","sexta","sabado","domingo")
print(type(dias))

print(f"Primeiro dia da semana é: {dias[0]}")
```
