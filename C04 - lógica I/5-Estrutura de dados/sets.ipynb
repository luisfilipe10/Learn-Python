sets são conjuntos de dados não ordenados únicos e imutáveis para realização de cálculos. (união e Interseção)

```python
#Criando um conjunto, o SET remove automáticamente as duplicatas

num = {1,1,3,3,4,4,4,5,6}
print(f"set 'num' sem duplicatas: {num}")
```

```python
#Adicionando um novo número

num.add(1001)
print(f"set com novo num add: {num}")
```

```python
#Adicionando mais de 1 número de uma vez.
num.update([5,6,7,8,9,10])
print(num)
```

```python
#Removendo número do set
num.remove(1001)
print(num)
```

```python
#Operação de conjuntos

#União -> Elementos adicionados como um único set

conjunto_a = {1,3,5,7,11,13}
conjunto_b = {1,3,6,9,13}

print(f"A união A/B: {conjunto_a.union(conjunto_b)}")



#Interseção -> Elementos presentes nos dois sets A = B

inter = conjunto_a.intersection(conjunto_b)
print(f"A interseção A/B: {inter}")
```
