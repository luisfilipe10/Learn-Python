Estrutura de dados : Listas

```python
carros = ["gol","corolla","civic","jac"]
print(carros)
type(carros)
```

```python
# Comando len para validar quantidade de registros na lista
len(carros)
```

```python
# comando [num] para acessar determinado item pelo índice na lista (0,-1)
print(f"lista completa: {carros}")
print(f"Primeiro item da lista: {carros[0]}")
print(f"último item da lista: {carros[-1]}")
```

```python
# Adicionando Item na lista (append, adiciona por último)

carros.append("palio")
print(carros)
```

```python
# Removendo Item na lista

carros.remove("corolla")
print(carros)
```

```python
# Removendo pelo Índice

carros.remove(carros[0])
print(carros)
```

```python
# adicionando carro na primeira posição  (0,-1)

print(carros)

carros.insert(0, "uno")
print(carros)
```
