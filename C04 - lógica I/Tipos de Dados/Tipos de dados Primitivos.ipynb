Tipos de Dados Primitivos em Python

Os tipos de dados primitivos são os tipos básicos usados para armazenar
informações em Python.

1. int --- Inteiro

Usado para representar números inteiros, positivos ou negativos.

idade = 25
quantidade = 10
saldo = -50

Exemplo:

numero = 100

print(numero)
print(type(numero))

Saída:

100
<class 'int'>

2. float --- Número decimal

Usado para representar números com casas decimais.

altura = 1.82
preco = 49.90
temperatura = -2.5

Exemplo:

preco = 19.99

print(preco)
print(type(preco))

Saída:

19.99
<class 'float'>

Em Python, números decimais utilizam . e não ,.

3. str --- Texto

Usado para armazenar textos e caracteres.

Pode ser escrito utilizando aspas simples ou duplas.

nome = "Luís"
cidade = 'João Monlevade'

Exemplo:

nome = "Carlos"

print(nome)
print(type(nome))

Saída:

Carlos
<class 'str'>

4. bool --- Booleano

Representa valores lógicos.

Possui apenas dois valores:

True
False

Exemplo:

maior_de_idade = True
tem_carteira = False

print(maior_de_idade)
print(type(maior_de_idade))

Saída:

True
<class 'bool'>

Operadores de comparação

Comparações normalmente retornam um bool.

idade = 20

print(idade >= 18)
print(idade == 20)
print(idade < 18)

Resultado:

True
True
False

Conversão de Tipos

Python permite converter valores entre diferentes tipos usando funções
como:

int()

float()

str()

bool()

str para int

idade = "25"

idade = int(idade)

print(idade)
print(type(idade))

int para float

numero = 10

numero = float(numero)

print(numero)

Resultado:

10.0

int para str

numero = 100

texto = str(numero)

print(texto)
print(type(texto))

NoneType --- Ausência de valor

Embora normalmente não seja tratado como um tipo primitivo tradicional,
None é muito importante em Python.

Ele representa a ausência de um valor.

resultado = None

print(resultado)
print(type(resultado))

Saída:

None
<class 'NoneType'>

É comum encontrar None em funções que não retornam explicitamente um
valor.

Resumo

Tipo         Nome                Exemplo

int        Inteiro             10
float      Decimal             10.5
str        Texto               "Python"
bool       Booleano            True
NoneType   Ausência de valor   None

Exemplo prático

nome = "Luís"
idade = 25
altura = 1.82
estuda_python = True
resultado = None

print(nome)
print(idade)
print(altura)
print(estuda_python)
print(resultado)
