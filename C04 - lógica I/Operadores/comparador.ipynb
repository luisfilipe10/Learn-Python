## Operadores de Comparação 

aqui temos operadores lógicos voltados para comparação de variáveis com 2 retornos possíveis (True or False)

```python
#Definindo Var

a = 20
b = 11
```

```python
# maior que 
print(f'{a} > {b}?  {a >= b}')
```

```python
# menor que 
print(f'{a} < {b}?  {a < b}')
```

```python
# Igual a 
print(f'{a} = {b}?  {a == b}')
```

```python
# Diferente
print(f'{a} != {b}?  {a != b}')
```
