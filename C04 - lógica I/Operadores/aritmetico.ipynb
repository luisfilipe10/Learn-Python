## Operadores Aritiméticos e comparadores Lógicos

```python
#Def de variáveis

a = 10
b = 3
```

```python
soma = a + b 
subtracao = a - b
multiplicacao = a * b
divisao = a / b
divisao_inteira = a // b
modulo = a % b
potencia = a ** b
```

```python
print(f'\nA soma é: {soma} ')
print(f'\nA subtração é: {subtracao} ')
print(f'\nA multiplicacao é: {multiplicacao} ')
print(f'\nA divisao é: {divisao} ')
print(f'\nA divisao_inteira é: {divisao_inteira} ')
print(f'\nO modulo é: {modulo} ')
print(f'\nA potência é: {potencia} ')
```
