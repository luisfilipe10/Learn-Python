 Operadores Lógicos (Resumo Simples)

Serve para juntar perguntas cujas respostas são **Sim** ou **Não** (`True` ou `False`).

---

 1. `E` (AND / &&)
* **Regra:** Tudo precisa ser **Sim**. Se um for **Não**, dá **Não**.
* **Exemplo:** Ter 18 anos **E** ter carteira de motorista. 
  * Se tiver idade mas não tiver carteira $\rightarrow$ **Não** pode dirigir.

 2. `OU` (OR / ||)
* **Regra:** Basta **um** ser **Sim** para dar **Sim**. Só dá **Não** se tudo for **Não**.
* **Exemplo:** Ser estudante **OU** ter cupom de desconto.
  * Se não for estudante mas tiver cupom $\rightarrow$ **Ganhou** desconto.

 3. `NÃO` (NOT / !)
* **Regra:** Inverte o valor. O que é **Sim** vira **Não**, e o que é **Não** vira **Sim**.
* **Exemplo:** `não logado`.
  * Se estiver logado (Sim), o `NÃO` transforma em **Falso**.

```python
nome = input("Insira teu nome: ")
idade = int(input("Insira tua idade: "))
carteira =input("Insira tua habilitação (A / B / C / D / E): ").upper()
categoria = ("A","B","C","D","E")

if carteira not in categoria or idade < 18:
    print(f"{nome}, não está apto")
else:
    print(f"Parabéns {nome}, você poderá seguir com a inscrição do curso !")
```
