## Operadores de Comparação 

aqui temos operadores lógicos voltados para comparação de variáveis com 2 retornos possíveis (True or False)

```python
#Definindo Var

a = 20
b = 11
```

```python
# maior que 
a > b
```

```python
# menor que 
a < b
```

```python
# Igual a 
a == b
```

```python
# Diferente
a != b
```
