String  (palavras)

sequência de caracteres com métodos úteis 

```python
#Manipulando uma frase, gerando padrão para Maiuscula e Minuscula


frase =" Python é uma linguagem de programação "
print(frase)
print(f"\n Frase maiuscula: {frase.upper()}")
print(f"\n Frase minuscula: {frase.lower()}")
frase_corrigida = frase.strip()
print(f"\n Frase corrigida, sem espaços em branco antes e depois: {frase_corrigida}")

```

```python
#Concatenação

nome = "Luis"
saudacao = "Olá"+ "," + " " +nome + "!"

print(saudacao) 
```

```python
#Len para obter a quantidade de caracteres da frase / string
frase =" Meu nome é Luís, e estou aprendendo Python" 
len(frase.strip())
```

```python
#Replace, para substituir determinada string

frase = "Belo Horizonte é a melhor capital do Brasil"
print(frase)
print(" ")
print(f"Substituindo a capital BH por Vitória: {frase.replace("Belo Horizonte","Vitória")}")
```

```python
#Slicing (Fatiamento), aqui buscamos partes de uma strig
#Em Python, o índice começa em 0

frase =" Python é uma linguagem"

print(frase)
print("")
print(f"\nQuantidade de caracteres: {len(frase)}")
print(f"\nPrimeiro caractere (0): {frase[0]}")
print(f"\nPrimeiro caractere (1): {frase[1]}")
print(f"\nCaracteres (3 ao 14) : {frase[3:14]}")

```
