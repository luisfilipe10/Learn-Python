Estrutura	Mutável / Imutável	O que pode ser alterado?	Observação
Lista (list)	Mutável	Conteúdo (elementos), tamanho, ordem	Pode adicionar, remover, substituir itens.
Dicionário (dict)	Mutável	Chaves e valores, tamanho, ordem (Python 3.7+)	Chaves devem ser hash‑able (imutáveis).
Tupla (tuple)	Imutável	Conteúdo (elementos)	Não pode ser alterado após criação; pode conter objetos mutáveis.

```python
#criando uma lista

frutas = ["maçã","manga","laranja","kiwi"]
```

```python
print(frutas)
print(f"\nPrimeira fruta é: {frutas[0]}")
print(f"\nÚltima fruta é: {frutas[-1]}")
```

```python
#Removendo laranja

frutas.remove("laranja")
print(frutas)
```

```python
#Adicionando Abacaxi

frutas.append("abacaxi")

print(frutas)
```

```python
#Modificando itens

print(frutas) #original
frutas[1] = "uva" #modificando
print(frutas) #modificado
```

```python
#Método Len() para ler qtd de itens na lista

print(f"\nA lista contém: {len(frutas)} frutas")
```
