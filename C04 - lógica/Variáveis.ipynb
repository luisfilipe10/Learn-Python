Conceitos Básicos de Python: Lógica, Variáveis e Tipos de Dados

Nesta seção, exploramos os três pilares fundamentais para iniciar a programação em Python.

1. Lógica de Programação

A lógica de programação é a sequência de passos utilizada para resolver um problema. Em Python, estruturamos essa lógica utilizando:

Estruturas Condicionais (if, elif, else): Permitem que o código tome decisões com base em condições booleanas (True ou False).

Estruturas de Repetição (for, while): Automatizam tarefas repetitivas, iterando sobre sequências ou executando blocos de código enquanto uma condição for verdadeira.

2. Variáveis

Variáveis são espaços na memória do computador usados para armazenar dados. Em Python, a declaração é dinâmica, ou seja, você não precisa declarar o tipo explicitamente; basta atribuir um valor usando o operador =.

# Exemplo de atribuição de variável
nome = "Ana"
idade = 25


3. Tipos de Dados Básicos - Python é uma Linguagem Dinamicamente Tipada

Python possui vários tipos de dados nativos para categorizar as informações que manipulamos. Os principais são:

Inteiros (int): Números inteiros, positivos ou negativos (ex: 42, -3).

Ponto Flutuante (float): Números reais, com casas decimais (ex: 3.14, -0.001).

Strings (str): Textos, delimitados por aspas simples ou duplas (ex: "Olá, mundo!").

Booleanos (bool): Valores lógicos que representam verdadeiro (True) ou falso (False).

Coleções (Listas, Tuplas e Dicionários): Estruturas para armazenar múltiplos valores de forma organizada.

## Váriaveis Globais VS Variáveis Locais

Uma Var fora do bloco de função def(), é uma Var Global, sendo reutilizável durante todo o código,
Já uma var Local é o oposto, sento utilizável apenas dentro da def(função)



```python
nome = 'Luis' #var global
saudacao = 'Olá, bem vindo ao LFMA' # var global

def minha_saudacao():
    nome = 'Arthur'
    print(f'\n Dentro da função {nome}')
    print(saudacao)
minha_saudacao()

print(f'\n fora da função, {saudacao},{nome}')
```

```python
imposto = 0.60
def calculadora():
    salario = int(input('Insira o seu salario: '))
    salario_final = salario -(salario * imposto)
    print(salario_final)
calculadora()
```
